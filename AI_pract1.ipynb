```python
#13. black and white noise caused during transmission. Write a Python program to add salt-and-pepper noise and remove it using a *median filter*. | Median Filtering |
# pip install opencv-python numpy matplotlib
import cv2
import numpy as np
import matplotlib.pyplot as plt

def load(path='input.jpg'):
    im = cv2.imread(path)
    if im is None:
        print(path, 'not found - using a demo image. Put your image next to the notebook as input.jpg')
        im = np.zeros((300, 400, 3), np.uint8); im[:] = (60, 40, 30)
        cv2.rectangle(im, (40, 40), (180, 160), (0, 0, 255), -1)
        cv2.circle(im, (290, 110), 60, (255, 120, 0), -1)
        cv2.rectangle(im, (100, 200), (350, 270), (0, 200, 0), -1)
    return im

def show(images, titles, cols=None):
    cols = cols or len(images)
    rows = (len(images) + cols - 1) // cols
    plt.figure(figsize=(5 * cols, 4 * rows))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(rows, cols, i + 1)
        if im.ndim == 2: plt.imshow(im, cmap='gray')
        else: plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        plt.title(t); plt.axis('off')
    plt.tight_layout(); plt.show()

img = load()
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

def add_salt_pepper(image, amount=0.05):
    noisy = image.copy()
    n = int(amount * image.size)
    ys = np.random.randint(0, image.shape[0], n); xs = np.random.randint(0, image.shape[1], n)
    noisy[ys[:n // 2], xs[:n // 2]] = 255      # salt
    noisy[ys[n // 2:], xs[n // 2:]] = 0        # pepper
    return noisy

noisy = add_salt_pepper(gray, 0.05)
cleaned = cv2.medianBlur(noisy, 3)
show([gray, noisy, cleaned], ['Original', 'Salt & Pepper noise', 'Median filtered'])
```

```python
#18.  Locate human faces in a photograph. Write a Python program using a suitable OpenCV face-detection method to detect faces and draw a rectangle around each detected face. | Face Detection |
# pip install opencv-python numpy matplotlib
import os
import glob
import urllib.request
import cv2
import numpy as np
import matplotlib.pyplot as plt

def show(images, titles, cols=None):
    cols = cols or len(images)
    plt.figure(figsize=(6 * cols, 5))
    for i, (im, t) in enumerate(zip(images, titles)):
        plt.subplot(1, cols, i + 1)
        plt.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        plt.title(t); plt.axis('off')
    plt.tight_layout(); plt.show()

def get_cascade_path(name='haarcascade_frontalface_default.xml'):
    # 0) already in the current folder
    if os.path.exists(name):
        return name
    # 1) normal OpenCV location (if cv2.data exists)
    data = getattr(cv2, 'data', None)
    if data is not None:
        p = os.path.join(data.haarcascades, name)
        if os.path.exists(p):
            return p
    # 2) search inside the cv2 package folder and common install folders
    folders = [os.path.dirname(cv2.__file__)]
    folders += glob.glob(os.path.join(os.path.dirname(os.__file__), 'site-packages', 'cv2'))
    for folder in folders:
        for root, _, files in os.walk(folder):
            if name in files:
                return os.path.join(root, name)
    # 3) download once into the current folder (needs internet)
    url = 'https://raw.githubusercontent.com/opencv/opencv/master/data/haarcascades/' + name
    urllib.request.urlretrieve(url, name)
    return name

img = cv2.imread('face.jpg')    # use a photo that contains faces
if img is None:
    raise FileNotFoundError('Put a photo with faces next to this notebook as face.jpg')

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cascade = cv2.CascadeClassifier(get_cascade_path())
if cascade.empty():
    raise RuntimeError('Could not load the Haar cascade file')

faces = cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

out = img.copy()
for (x, y, w, h) in faces:
    cv2.rectangle(out, (x, y), (x + w, y + h), (0, 255, 0), 2)

print('Faces detected:', len(faces))
show([img, out], ['Original', f'Detected faces: {len(faces)}'])
```

```python
#Practical 9 - Edge Detection (Canny)
# What it does: Opens an image with Pillow, converts it to a NumPy array and then to grayscale
# (handles gray, RGB and RGBA images), applies Canny edge detection
# (thresholds 100 and 200), shows original vs edges and saves edge_detected.jpg.
 
# pip install opencv-python pillow numpy matplotlib
from PIL import Image
import cv2
import numpy as np
import matplotlib.pyplot as plt
 
# Open image using Pillow
img = Image.open("in.png")
 
# Convert image to NumPy array
img_np = np.array(img)
 
# Check image channels and convert to grayscale if needed
if len(img_np.shape) == 2:
    # Image is already grayscale
    gray = img_np
elif img_np.shape[2] == 4:
    # RGBA image
    gray = cv2.cvtColor(img_np, cv2.COLOR_RGBA2GRAY)
else:
    # RGB image
    gray = cv2.cvtColor(img_np, cv2.COLOR_RGB2GRAY)
 
# Apply Canny Edge Detection
edges = cv2.Canny(gray, 100, 200)
 
# Display Original Image
plt.figure(figsize=(12, 6))
 
plt.subplot(1, 2, 1)
if len(img_np.shape) == 2:
    plt.imshow(img_np, cmap='gray')
else:
    plt.imshow(img_np)
plt.title("Original Image")
plt.axis("off")
 
# Display Edge Detected Image
plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title("Edge Detected Image")
plt.axis("off")
 
plt.tight_layout()
plt.show()
 
# Save edge-detected image
cv2.imwrite("edge_detected.jpg", edges)
 
print("Edge detection completed successfully!")
print("Output saved as 'edge_detected.jpg'")
 
```
